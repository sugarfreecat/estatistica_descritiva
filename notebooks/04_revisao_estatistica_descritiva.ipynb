# Revisão de Estatística Descritiva

## Funções Úteis

- Média (mean)
- Mediana (median)
- Moda (mode)
- Amplitude (max-min)
- Desvio Padrão (std)
- Correlação (corr)

### Média (mean):
Imagine que você e 4 amigos ganharam quantidades diferentes de balas. Um ganhou 2, outro 8, outro 10... A média é o ato de juntar todas as balas em um único monte e dividir exatamente por igual entre todo mundo.

### Mediana (median):
Imagine que você colocou todas as pessoas da sala em uma fila indiana, organizadas da mais baixa para a mais alta. A mediana é exatamente a pessoa que está de pé bem no meio da fila de mãos dadas.

### Moda (median):
É o número ou números que mais se repetem; remetem à moda - tendência - do momento.

### Amplitude (max-min):
É como esticar uma fita métrica desde o menor número até o maior número de um grupo. Qual é a distância entre o menor e o maior valor?

### Desvio Padrão (std)
Imagine uma professora (que é a média) tentando guiar um grupo de crianças no parquinho. Se todas as crianças estão coladinhas na saia da professora, o desvio padrão é baixinho (todo mundo comportado e pertinho do centro).

Se as crianças estão correndo para os quatro cantos do parquinho (uma no escorregador, uma no balanço, uma no lanche), o desvio padrão é altíssimo (muita bagunça e espalhamento)

Ele mede o quanto os dados estão "fugindo" da média. Ajuda a saber se a situação é estável ou uma montanha-russa.

### Correlação (corr)
É o "Match do Tinder" entre duas colunas de números.

Se quanto mais você estuda (Coluna A), mais sua nota aumenta (Coluna B), eles dão match perfeito: a correlação é positiva e forte (vai até +1)

Se quanto mais frio está lá fora (Coluna A), menos sorvetes você vende (Coluna B), um sobe e o outro desce: a correlação é negativa (vai até -1)

Se não tem nada a ver uma coisa com a outra (ex: número de calçado versus quantidade de chuva), a correlação é zero.

#### CORRELAÇÃO POSITIVA - VÃO PARA O MESMO LADO
Quando uma coisa aumenta, a outra também aumenta. Se uma coisa diminui a outra também diminui.

#### CORRELAÇÃO NEGATIVA - VÃO PARA LADOS OPOSTOS
Quando uma coisa aumenta, a outra diminui. Gangorra

#### CORRELAÇÃO ZERO - NÃO TEM NADA A VER
Quando uma coisa muda, a outra não está nem aí. Não existe nenhum padrão entre elas. Mudar um não faz diferença nenhuma no outro.

## Gráficos


### 1. Gráfico de Linhas
Mostra evolução/tendência ao longo do tempo. Ex: vendas mês a mês.


### 2. Gráfico de Barras/Colunas
Usado para comparar valores entre categorias diferentes. Ex: vendas do produto x vs o produto y.


### 3. Gráfico de Pizza
Usado para mostrar proporções (%) em poucas fatias. Ex: divisão de sócios em uma empresa.


### 4. Box Plot (Caixa)
Usado para mostrar mediana, dispersão e outliers. Ex: comparar notas entre equipes.


### 5. Dispersão
Usado para mostrar a correlação entre duas variáveis numéricas. Ex: investimento em anúncios vs vendas.

# Cenário da Cafeteria Gen

```python
# importar bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# criar nossa base de dados simples
dados = {
    "dia": ["segunda", "terça", "quarta", "quinta", "sexta", "sábado", "domingo"],
    "cafes_vendidos": [50, 55, 50, 60, 65, 180, 70],
    "temperatura": [22, 20, 21, 19, 18, 15, 17]
}

df = pd.DataFrame(dados)
print(df)
display(df.describe().round(1))

sns.scatterplot(
    data=df,
    x='temperatura',       # Eixo X: Temperatura (quanto mais à direita, mais calor)
    y='cafes_vendidos',    # Eixo Y: Cafés vendidos (quanto mais para cima, mais vendas)
    hue='dia',   # Cor dos pontos: Muda de acordo com o dia da semana
    s=120,                 # Tamanho dos pontos para ficarem fáceis de ver
    palette='deep'         # Paleta de cores agradável
)

# 3. Adicionando títulos e melhorando a estética
plt.title('Relação entre Temperatura, Vendas de Café e Dias da Semana', fontsize=14, fontweight='bold')
plt.xlabel('Temperatura (°C)', fontsize=12)
plt.ylabel('Cafés Vendidos (unidades)', fontsize=12)

# Jogando a legenda para fora do gráfico para não tampar os pontos
plt.legend(title='Dia da Semana', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adicionando linhas de grade quadriculada para facilitar a leitura
plt.grid(True, linestyle='--', alpha=0.5)

# 4. Exibindo o gráfico limpo
plt.tight_layout()
plt.show()
```

**Mini-Case 1: O Mistério do Bônus de Vendas (Média vs. Mediana)**

Cenário: Uma loja tem 5 vendedores. Em outubro, eles venderam: R$ 10.000, R$ 11.000, R$ 12.000, R$ 11.000 e R$ 56.000 (um vendedor fechou um contrato gigante único).
O gerente calculou a média (R$ 20.000) e quer exigir que todos vendam R$ 20.000 no mês seguinte.


*Como analista de dados, você concorda com essa meta? Qual medida
você mostraria ao gerente e qual é o valor dela?*

```python
df_vendas = pd.DataFrame({
    "vendas": [10000, 11000, 12000, 11000, 56000]
})

display(df_vendas)

vendas = df_vendas["vendas"]

print(f"Média de vendas: {vendas.mean()}")
print(f"Mediana de vendas: {vendas.median()}")
print(f"Amplitude de vendas: {vendas.max() - vendas.min()}")
print(f"Q1: {vendas.quantile(0.25)}")
print(f"Q2: {vendas.quantile(0.50)}")
print(f"Q3: {vendas.quantile(0.75)}")
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot( y=vendas, color='lightcoral', width=0.4 )

plt.title('Análise de Vendas da Equipe: O Impacto do Outlier', fontsize=12, fontweight='bold')
plt.ylabel('Valor de Vendas (R$)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()

plt.show()
```

### Conclusão
Não acredito que definir a meta de vendas para R$20.000 apenas com base na média seja uma boa ideia.

Para convencer o gerente, eu mostraria primeiro a mediana, no valor de R$11.000, para demonstrar que os dados não estão distribuídos uniformemente: há algum outlier distorcendo o valor da média para acima da mediana.

Além disso, acredito que os quartis também sejam bons indicadores da distorção da média: o terceiro quartil tem o valor de apenas R$12.000; ou seja, até 75% dos vendedores só venderam até esse valor. Logo, não é realista esperar que eles comecem a vender muito a mais que isso.

Por último, para reforçar a ideia de que há um outlier distorcendo a média e que ele não deve representar as vendas de todos os outros vendedores, eu mostraria três indicadores:

1. A prova de que há realmente um outlier. (fazer dps)
2. A amplitude entre o menor e o maior valor (46000)
3. O gráfico acima como um facilitador visual. Nele, é possível perceber que todos os vendedores tiveram vendas entre 10 e 12 mil reais, exceto uma única pessoa, com 56 mil reais.
